```python
# SESSION 9 - Working with Date & Time Data


# Task 1
# Given a list of delivery timestamps as strings, use pandas and
# pd.to_datetime() to convert them into datetime objects and print the result.

import pandas as pd
import numpy as np

delivery_timestamps = [
    "2024-06-01 14:30",
    "2024-06-02 09:15",
    "2024-06-03 20:45"
]

delivery_dates = pd.to_datetime(delivery_timestamps)

print("Delivery Timestamps:")
print(delivery_dates)

```

```python

# Task 2
# Load a CSV file containing order dates from a Flipkart-style order history
# and extract the year, month, and weekday for each order.

order_data = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "order_date": [
        "2024-06-01 14:30:00",
        "2024-06-02 09:15:00",
        "2024-06-03 20:45:00",
        "2024-06-08 12:10:00",
        "2024-06-10 18:30:00",
        "2024-06-15 11:20:00",
        "2024-06-17 16:45:00"
    ]
})

order_data["order_date"] = pd.to_datetime(
    order_data["order_date"]
)

order_data["year"] = order_data["order_date"].dt.year
order_data["month"] = order_data["order_date"].dt.month
order_data["weekday"] = order_data["order_date"].dt.day_name()

print("\nOrder Data with Date Features:")
print(order_data)
```

```python
# Task 3
# Set the 'order_date' column as the index and use resampling
# to calculate the total number of orders placed each week.

order_data = order_data.set_index("order_date")

weekly_orders = order_data.resample("W").size()

print("\nWeekly Order Count:")
print(weekly_orders)

```

```python

# Task 4
# Suppose you have a DataFrame of Instagram posts with a 'posted_at'
# column in UTC. Convert these timestamps to 'Asia/Kolkata' timezone
# and display the first 5 converted times.

instagram_data = pd.DataFrame({
    "post_id": [1, 2, 3, 4, 5, 6],
    "posted_at": [
        "2024-06-01 10:00:00",
        "2024-06-02 12:30:00",
        "2024-06-03 15:45:00",
        "2024-06-04 08:20:00",
        "2024-06-05 18:10:00",
        "2024-06-06 20:00:00"
    ]
})

instagram_data["posted_at"] = pd.to_datetime(
    instagram_data["posted_at"],
    utc=True
)

instagram_data["posted_at"] = instagram_data["posted_at"].dt.tz_convert(
    "Asia/Kolkata"
)

print("\nInstagram Posts - India Time:")
print(instagram_data["posted_at"].head())

```

```python

# Task 5
# Create a new feature called 'is_weekend' that marks True if
# an order was placed on Saturday or Sunday, and False otherwise.

order_data["is_weekend"] = order_data.index.dayofweek >= 5

print("\nOrders with Weekend Feature:")
print(order_data[["order_id", "is_weekend"]])
```
