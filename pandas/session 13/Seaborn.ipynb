```python
# SESSION 13 - Data Visualization with Seaborn (Part 2)


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# QUESTION 1
# Load the tips dataset and create a pairplot
# for all numeric variables.

tips = sns.load_dataset("tips")

sns.pairplot(
    tips
)

plt.show()

```

```python
# QUESTION 2
# Load the flights dataset and create a heatmap
# showing passengers by month and year.

flights = sns.load_dataset("flights")

flight_table = flights.pivot_table(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(
    flight_table,
    annot=True,
    fmt=".0f"
)

plt.title("Passengers by Month and Year")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```

```python

# QUESTION 3
# Use relplot with the fmri dataset to visualize
# signal changes over time for different event types.

fmri = sns.load_dataset("fmri")

sns.relplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    kind="line"
)

plt.show()


```

```python
# QUESTION 4
# Use catplot with the titanic dataset to display
# survival rate for each passenger class.

titanic = sns.load_dataset("titanic")

sns.catplot(
    data=titanic,
    x="class",
    y="survived",
    kind="bar",
    errorbar="ci"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()


```

```python
# QUESTION 5
# Use jointplot to show the relationship between
# bill length and flipper length with regression line.

penguins = sns.load_dataset("penguins")

sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="flipper_length_mm",
    kind="reg"
)

plt.show()
```
