```python
# SESSION 14 - Exploratory Data Analysis (EDA) (Part 1)


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# QUESTION 1
# Load IPL match dataset and perform univariate analysis
# on the total_runs column.

ipl_data = pd.read_csv("ipl_matches.csv")

print("Total Runs Summary:")

print("Mean:", ipl_data["total_runs"].mean())
print("Median:", ipl_data["total_runs"].median())
print("Minimum:", ipl_data["total_runs"].min())
print("Maximum:", ipl_data["total_runs"].max())
print("Standard Deviation:", ipl_data["total_runs"].std())

```

```python

# QUESTION 2
# Load Flipkart product reviews and create a bar plot
# showing the number of reviews for each rating.

flipkart_data = pd.read_csv("flipkart_reviews.csv")

rating_counts = flipkart_data["rating"].value_counts().sort_index()

sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values
)

plt.title("Number of Reviews by Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")

plt.show()

```

```python

# QUESTION 3
# Load Zomato restaurant data and create a scatterplot
# showing the relationship between cost and user rating.

zomato_data = pd.read_csv("zomato_restaurants.csv")

sns.scatterplot(
    data=zomato_data,
    x="average_cost_for_two",
    y="user_rating"
)

plt.title("Restaurant Cost vs User Rating")
plt.xlabel("Average Cost for Two")
plt.ylabel("User Rating")

plt.show()

```

```python

# QUESTION 4
# Load BookMyShow movie data and find the average
# box office collection for each genre.

movies_data = pd.read_csv("bookmyshow_movies.csv")

genre_collection = (
    movies_data
    .groupby("genre")["box_office_collection"]
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage Box Office Collection by Genre:")
print(genre_collection)

```

```python

# QUESTION 5
# Load Spotify songs dataset and create a pairplot
# for danceability, energy, valence and popularity.

spotify_data = pd.read_csv("spotify_songs.csv")

spotify_features = spotify_data[
    [
        "danceability",
        "energy",
        "valence",
        "popularity"
    ]
]

sns.pairplot(
    spotify_features
)

plt.show()


# One interesting pattern:
# Songs with higher energy often show higher danceability,
# meaning energetic songs are frequently more suitable for dancing.
```
