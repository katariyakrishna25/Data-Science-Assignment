```python
# SESSION 12 - Data Visualization with Seaborn (Part 1)


import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# QUESTION 1
# Create a histplot showing delivery time distribution
# for 50 Zomato food orders.


np.random.seed(42)

delivery_times = np.random.randint(15, 61, 50)

sns.histplot(
    delivery_times,
    bins=8,
    kde=True
)

plt.title("Zomato Delivery Time Distribution")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.show()


```

```python
# QUESTION 2
# Load the tips dataset and create a boxplot
# showing total bill amounts by day.


tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Total Bill Amount by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")

plt.show()


```

```python
# QUESTION 3
# Simulate IPL scores for 8 teams
# and compare their run distributions using violinplot.


np.random.seed(42)

teams = [
    "CSK",
    "MI",
    "RCB",
    "KKR",
    "GT",
    "RR",
    "SRH",
    "DC"
]

ipl_data = []

for team in teams:
    scores = np.random.randint(120, 221, 10)

    for score in scores:
        ipl_data.append(
            {
                "Team": team,
                "Runs": score
            }
        )

ipl_df = pd.DataFrame(ipl_data)


sns.set_theme(style="darkgrid")

sns.violinplot(
    data=ipl_df,
    x="Team",
    y="Runs"
)

plt.title("IPL Run Distribution by Team")
plt.xlabel("Team")
plt.ylabel("Runs")

plt.show()


```

```python
# QUESTION 4
# Create a countplot showing the number of songs
# per genre from 40 Spotify tracks.


genres = [
    "Pop", "Rock", "Hip-Hop", "Bollywood",
    "Pop", "Rock", "Pop", "Hip-Hop",
    "Bollywood", "Rock", "Pop", "Hip-Hop",
    "Bollywood", "Pop", "Rock", "Pop",
    "Hip-Hop", "Bollywood", "Rock", "Pop",
    "Pop", "Rock", "Hip-Hop", "Bollywood",
    "Pop", "Rock", "Hip-Hop", "Pop",
    "Bollywood", "Rock", "Pop", "Hip-Hop",
    "Bollywood", "Pop", "Rock", "Pop",
    "Hip-Hop", "Bollywood", "Rock", "Pop"
]

spotify_df = pd.DataFrame(
    {
        "Genre": genres
    }
)


sns.countplot(
    data=spotify_df,
    x="Genre"
)

plt.title("Number of Spotify Songs by Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")

plt.show()


```

```python
# QUESTION 5
# Use kdeplot to visualize daily step counts
# for a fitness app.


daily_steps = [
    5200,
    6800,
    7500,
    6200,
    8100,
    9000,
    7200
]


sns.set_theme(style="whitegrid")

sns.kdeplot(
    daily_steps,
    fill=True
)

plt.title("Daily Step Count Distribution")
plt.xlabel("Steps")
plt.ylabel("Density")

plt.show()
```
