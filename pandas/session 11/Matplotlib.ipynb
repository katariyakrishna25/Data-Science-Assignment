```python
# SESSION 11 - Data Visualization: Matplotlib (Part 2)


import matplotlib.pyplot as plt
import numpy as np


# QUESTION 1
# Create a 2x2 grid of subplots with line, bar, scatter and pie charts.


fig, axes = plt.subplots(2, 2, figsize=(12, 8))


# Line Chart

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
orders = [120, 150, 130, 180, 200]

axes[0, 0].plot(
    days,
    orders,
    marker="o",
    linewidth=2
)

axes[0, 0].set_title("Daily Orders")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Orders")
axes[0, 0].grid(True)


# Bar Chart

categories = ["Indian", "Pizza", "Chinese", "Fast Food"]
restaurant_count = [25, 18, 12, 20]

axes[0, 1].bar(
    categories,
    restaurant_count
)

axes[0, 1].set_title("Restaurants by Category")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Number of Restaurants")


# Scatter Chart

delivery_time = [20, 25, 30, 35, 40, 45]
customer_rating = [4.8, 4.5, 4.3, 4.0, 3.8, 3.5]

axes[1, 0].scatter(
    delivery_time,
    customer_rating,
    marker="o"
)

axes[1, 0].set_title("Delivery Time vs Rating")
axes[1, 0].set_xlabel("Delivery Time (minutes)")
axes[1, 0].set_ylabel("Customer Rating")


# Pie Chart

payment_methods = ["UPI", "Card", "Cash", "Wallet"]
payment_orders = [40, 25, 20, 15]

axes[1, 1].pie(
    payment_orders,
    labels=payment_methods,
    autopct="%1.1f%%"
)

axes[1, 1].set_title("Payment Method Distribution")


plt.tight_layout()
plt.show()

```

```python

# QUESTION 2
# Compare average delivery times for Zomato, Swiggy and Domino's.
# Each bar uses different styling options.


platforms = ["Zomato", "Swiggy", "Domino's"]
delivery_times = [32, 35, 28]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    platforms,
    delivery_times,
    color=["red", "orange", "blue"],
    edgecolor="black",
    linewidth=2
)

plt.title("Average Delivery Time")
plt.xlabel("Food Delivery Platform")
plt.ylabel("Average Delivery Time (Minutes)")

plt.grid(axis="y", linestyle="--", linewidth=0.5)

plt.show()

```

```python

# QUESTION 3
# Multi-axis chart showing Instagram followers
# and average daily posts for five influencers.


influencers = [
    "Influencer A",
    "Influencer B",
    "Influencer C",
    "Influencer D",
    "Influencer E"
]

followers = [120000, 250000, 180000, 320000, 400000]
daily_posts = [2, 4, 3, 5, 6]


fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(
    influencers,
    followers,
    marker="o",
    linewidth=2
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Instagram Followers")
ax1.tick_params(axis="y")


ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    marker="s",
    linewidth=2
)

ax2.set_ylabel("Average Daily Posts")
ax2.tick_params(axis="y")


plt.title("Instagram Followers and Average Daily Posts")

plt.show()

```

```python

# QUESTION 4
# Plot tickets sold versus IMDb ratings.
# Add movie names above each point.


movies = [
    "Jawan",
    "Pathaan",
    "Animal",
    "Stree 2",
    "Dangal"
]

tickets_sold = [950, 850, 780, 720, 900]
imdb_ratings = [6.9, 5.8, 6.1, 6.9, 8.3]


plt.figure(figsize=(10, 6))

plt.scatter(
    tickets_sold,
    imdb_ratings,
    marker="o"
)

for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], imdb_ratings[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Movie Tickets Sold vs IMDb Ratings")
plt.xlabel("Tickets Sold")
plt.ylabel("IMDb Rating")

plt.grid(True)

plt.show()


```

```python
# QUESTION 5
# Plot Flipkart monthly sales and mark
# the highest sales point as Big Billion Days.


months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June"
]

sales = [120, 150, 135, 180, 160, 300]


fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    months,
    sales,
    marker="o",
    linewidth=2
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (in Lakhs)")

highest_index = sales.index(max(sales))

ax.annotate(
    "Big Billion Days",
    xy=(months[highest_index], sales[highest_index]),
    xytext=(-60, 30),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

ax.grid(True)

plt.show()
```
