```python
# SESSION 8 - Data Cleaning (Part 3: Duplicates, Typos, Consistency)


# Task 1
# Download a small sample of your recent Zomato order history as a CSV
# (or create a mock CSV with restaurant names and order dates), then use
# pandas' duplicated() function to find and print any duplicate orders
# based on restaurant name and date.

import pandas as pd

zomato_data = pd.DataFrame({
    "restaurant_name": [
        "Spice Hub",
        "Food Point",
        "Pizza House",
        "Spice Hub",
        "Burger King",
        "Food Point"
    ],
    "order_date": [
        "2026-09-01",
        "2026-09-02",
        "2026-09-03",
        "2026-09-01",
        "2026-09-04",
        "2026-09-02"
    ]
})

duplicates = zomato_data.duplicated(
    subset=["restaurant_name", "order_date"],
    keep=False
)

print("Duplicate Orders:")
print(zomato_data[duplicates])

```

```python
# Task 2
# Given a list of Flipkart product reviews with some duplicate entries,
# use value_counts() in pandas to identify which review texts are repeated
# most often and display the top 3 most common duplicate reviews.

reviews = pd.Series([
    "Good product",
    "Very useful product",
    "Good product",
    "Excellent quality",
    "Good product",
    "Average product",
    "Very useful product",
    "Excellent quality",
    "Good product",
    "Nice product"
])

review_counts = reviews.value_counts()

print("\nTop 3 Most Common Reviews:")
print(review_counts.head(3))
```

```python
# Task 3
# Create a DataFrame with mock data for Spotify playlists, including
# playlist names and creator usernames, where some rows are exact duplicates.
# Use drop_duplicates() to remove duplicate playlists and print the cleaned DataFrame.

spotify_data = pd.DataFrame({
    "playlist_name": [
        "Daily Hits",
        "Workout Songs",
        "Chill Vibes",
        "Daily Hits",
        "Party Mix",
        "Workout Songs"
    ],
    "creator_username": [
        "krishna",
        "rahul",
        "riya",
        "krishna",
        "neha",
        "rahul"
    ]
})

print("\nSpotify Playlists Before Removing Duplicates:")
print(spotify_data)

spotify_cleaned = spotify_data.drop_duplicates()

print("\nSpotify Playlists After Removing Duplicates:")
print(spotify_cleaned)

```

```python
# Task 4
# Suppose you have a DataFrame of Instagram usernames where some entries
# have typos like 'insta_queen', 'insta-queen', 'instaqueen'.
# Use replace() to standardize all these variants to 'instaqueen'.

instagram_data = pd.DataFrame({
    "username": [
        "insta_queen",
        "insta-queen",
        "instaqueen",
        "insta_queen",
        "insta-queen",
        "krishna123"
    ]
})

instagram_data["username"] = instagram_data["username"].replace({
    "insta_queen": "instaqueen",
    "insta-queen": "instaqueen"
})

print("\nStandardized Instagram Usernames:")
print(instagram_data)
```

```python
# Task 5
# You have a DataFrame column for payment status from a Paytm-like app
# with mixed values: 'Yes', 'yes', 'Y', 'No', 'no', 'N', and some with
# extra spaces. Write code to unify this column so all paid statuses become 1
# and all unpaid statuses become 0, trimming whitespace and fixing capitalization.

paytm_data = pd.DataFrame({
    "payment_status": [
        "Yes",
        " yes ",
        "Y",
        "No",
        " no ",
        "N",
        "YES",
        "  y  ",
        "NO"
    ]
})

paytm_data["payment_status"] = (
    paytm_data["payment_status"]
    .str.strip()
    .str.lower()
    .replace({
        "yes": 1,
        "y": 1,
        "no": 0,
        "n": 0
    })
)

print("\nCleaned Payment Status:")
print(paytm_data)
```
