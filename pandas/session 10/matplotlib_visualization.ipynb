```python
# SESSION 10 - Data Visualization: Matplotlib (Part 1)


# Task 1
# Install Matplotlib in your Python environment and create a simple line
# plot showing the number of daily steps taken over the last 7 days.
# Save the chart as steps_lineplot.png.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [6500, 8200, 7500, 9000, 7800, 10500, 8500]

plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker="o", label="Daily Steps")

plt.title("Daily Steps")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.legend()

plt.savefig("steps_lineplot.png")
plt.show()

```

```python

# Task 2
# Create a scatter plot of 10 restaurants with their Zomato ratings
# on the x-axis and average meal price on the y-axis.
# Add axis labels and a title.

restaurants = [
    "Spice Hub",
    "Food Point",
    "Pizza House",
    "Burger King",
    "Tasty Corner",
    "Urban Cafe",
    "Royal Bites",
    "Food Villa",
    "Green Bowl",
    "The Kitchen"
]

ratings = [4.5, 4.1, 4.6, 3.9, 4.3, 4.0, 4.7, 4.2, 4.4, 3.8]

meal_price = [450, 300, 550, 250, 400, 350, 600, 320, 380, 280]

plt.figure(figsize=(8, 5))
plt.scatter(ratings, meal_price)

plt.title("Restaurant Ratings vs Average Meal Price")
plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price")

plt.show()

```

```python

# Task 3
# Build a bar chart showing the number of orders placed on Swiggy,
# Zomato, and Domino's in the last month.
# Use different colors for each bar and add a legend.

apps = ["Swiggy", "Zomato", "Domino's"]
orders = [35, 42, 28]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    apps,
    orders,
    color=["orange", "red", "blue"],
    label="Orders"
)

plt.title("Food Delivery Orders")
plt.xlabel("Food Delivery App")
plt.ylabel("Number of Orders")
plt.legend()

plt.show()
```

```python

# Task 4
# Create a histogram of the durations of 20 Spotify listening sessions.
# Set the number of bins to 5 and customize the color of the bars.

listening_time = [
    15, 22, 35, 40, 18,
    25, 30, 45, 50, 20,
    28, 32, 38, 42, 55,
    17, 24, 36, 48, 60
]

plt.figure(figsize=(8, 5))

plt.hist(
    listening_time,
    bins=5,
    color="green",
    edgecolor="black"
)

plt.title("Spotify Listening Session Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Sessions")

plt.show()
```

```python
# Task 5
# Create a figure with two subplots:
# one line plot showing daily Instagram screen time for a week,
# and one bar chart showing the number of posts liked each day.
# Add titles and axis labels and save the figure as social_media_usage.png.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

screen_time = [2.5, 3.0, 2.8, 3.5, 3.2, 4.5, 4.0]

liked_posts = [12, 15, 10, 18, 14, 25, 20]

fig, axes = plt.subplots(2, 1, figsize=(9, 10))

axes[0].plot(
    days,
    screen_time,
    marker="o",
    label="Screen Time"
)

axes[0].set_title("Daily Instagram Screen Time")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Screen Time (Hours)")
axes[0].legend()


axes[1].bar(
    days,
    liked_posts,
    color="purple",
    label="Liked Posts"
)

axes[1].set_title("Posts Liked Each Day")
axes[1].set_xlabel("Day")
axes[1].set_ylabel("Number of Posts")
axes[1].legend()

plt.tight_layout()

plt.savefig("social_media_usage.png")
plt.show()
```
