```python
# SESSION 15 - Exploratory Data Analysis (EDA) (Part 2: Auto-EDA + Case Study Setup)


import pandas as pd


# QUESTION 1
# Load Spotify Top 100 Songs dataset and generate
# an automatic EDA profile report.

from ydata_profiling import ProfileReport

spotify_data = pd.read_csv("spotify_top_100.csv")

spotify_profile = ProfileReport(
    spotify_data,
    title="Spotify Top 100 Songs EDA Report"
)

spotify_profile.to_file("spotify_profile_report.html")

print("Spotify EDA report created successfully.")

print("\nMissing Values:")
print(spotify_data.isnull().sum())

print("\nData Types:")
print(spotify_data.dtypes)

```

```python

# QUESTION 2
# Compare Zomato restaurant data for Mumbai and Delhi
# using Sweetviz.

import sweetviz as sv

mumbai_data = pd.read_csv("zomato_mumbai.csv")
delhi_data = pd.read_csv("zomato_delhi.csv")

zomato_comparison = sv.compare(
    [mumbai_data, "Mumbai"],
    [delhi_data, "Delhi"]
)

zomato_comparison.show_html(
    "zomato_mumbai_vs_delhi.html"
)

print("Mumbai vs Delhi comparison report created successfully.")

```

```python

# QUESTION 3
# Load Myntra product listings and explore the data
# using D-Tale.
# Filter products with price above Rs. 2000.

import dtale

myntra_data = pd.read_csv("myntra_products.csv")

print("Myntra Product Data:")
print(myntra_data.head())

myntra_dtable = dtale.show(myntra_data)

products_above_2000 = myntra_data[
    myntra_data["price"] > 2000
]

print("\nProducts with price above Rs. 2000:")
print(products_above_2000)

print("\nNumber of Products above Rs. 2000:")
print(len(products_above_2000))

```
