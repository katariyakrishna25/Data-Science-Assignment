```python
# session 5 :

# Task 1: Create playlist_ids with 5 song IDs
 
playlist_ids = [101, 202, 303, 404, 505]
print(playlist_ids)
```

```python
# Task 2: Add IDs using append() and extend()

playlist_ids.append(606)
playlist_ids.extend([707, 808])
print(playlist_ids)
```

```python
# Task 3: Remove last song using pop()

removed = playlist_ids.pop()
print("Removed:", removed)
print("Remaining:", playlist_ids)
```
